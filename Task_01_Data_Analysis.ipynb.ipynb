```python

# Load the dataset and view records

import pandas as pd
from google.colab import files

# Upload your CSV file
uploaded = files.upload()

# Load the CSV file into a DataFrame
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

# Display the first 5 records
print("First 5 Records:")
display(df.head(5))

# Display the last 5 records
print("Last 5 Records:")
display(df.tail(5))

# Display a random sample of 5 records
print("Random Sample of 5 Records:")
display(df.sample(n=5, random_state=42))
```

```python

# Dataset Information and Shape

# 1. Display dataset information
print("DATASET INFORMATION")
df.info()

# 2. Display number of rows and columns
rows, columns = df.shape

print("\nDATASET SHAPE")
print("Number of rows:", rows)
print("Number of columns:", columns)

# 3. Display all column names
print("\nCOLUMN NAMES")
print(df.columns.tolist())

# 4. Display data types of all columns
print("\nDATA TYPES")
print(df.dtypes)

# 5. Display non-null counts for each column
print("\nNON-NULL COUNTS")
print(df.notnull().sum())

# 6. Display missing values in each column
print("\nMISSING VALUES PER COLUMN")
print(df.isnull().sum())

# 7. Display total missing values
print("\nTOTAL MISSING VALUES")
print(df.isnull().sum().sum())
```

```python

# Identify Column Types

# 1. Identify numerical columns
numerical_columns = df.select_dtypes(include=["number"]).columns.tolist()

# 2. Identify categorical columns
categorical_columns = df.select_dtypes(include=["object", "category", "string"]).columns.tolist()

# 3. Identify date/time columns
datetime_columns = df.select_dtypes(include=["datetime", "datetimetz"]).columns.tolist()

# Display the results
print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)

print("\nDate/Time Columns:")
if datetime_columns:
    print(datetime_columns)
else:
    print("No date/time columns found.")
```

```python

# Step 6: Check and Handle Missing Values

import pandas as pd
import numpy as np

# 1. Count missing values per column
missing_count = df.isnull().sum()

# 2. Calculate missing value percentage per column
missing_percentage = (df.isnull().sum() / len(df)) * 100

# 3. Display missing values report
missing_report = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage (%)": missing_percentage.round(2)
})

print("Missing Values Report:")
display(missing_report)

# 4. Handle missing values
for column in df.columns:

    if df[column].isnull().any():

        # Fill numerical columns with median
        if pd.api.types.is_numeric_dtype(df[column]):
            df[column] = df[column].fillna(df[column].median())

        # Fill categorical columns with mode
        else:
            mode_values = df[column].mode(dropna=True)

            if not mode_values.empty:
                df[column] = df[column].fillna(mode_values.iloc[0])
            else:
                print(f"Column '{column}' contains only missing values.")

# 5. Display missing values after handling
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python

# Step 7: Remove Duplicate Rows

# 1. Count duplicate rows
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

# 2. Remove duplicate rows
df = df.drop_duplicates()

# 3. Reset the index after removing duplicates
df = df.reset_index(drop=True)

# 4. Verify that no duplicates remain
remaining_duplicates = df.duplicated().sum()

print("Duplicate rows after removal:", remaining_duplicates)
print("Total rows after cleaning:", len(df))

if remaining_duplicates == 0:
    print("Success: No duplicate rows remain.")
```

```python

# Step 8: Fix Data Types

import pandas as pd
import numpy as np

# 1. Display data types before conversion
print("Data Types Before Conversion:")
print(df.dtypes)

# 2. Convert numeric-looking text columns into numbers
for column in df.select_dtypes(include=["object", "string"]).columns:
    cleaned = df[column].astype("string").str.strip()

    # Remove common formatting from numeric values, such as commas
    numeric_text = cleaned.str.replace(",", "", regex=False)
    numeric_text = numeric_text.str.replace(r"[$£€]", "", regex=True)

    converted = pd.to_numeric(numeric_text, errors="coerce")

    # Convert only if all non-missing original values are numeric
    if cleaned.notna().sum() > 0 and (
        converted.notna().sum() == cleaned.notna().sum()
    ):
        df[column] = converted

# 3. Detect columns whose names suggest they contain dates or times
for column in df.columns:
    if any(word in column.lower() for word in ["date", "datetime", "timestamp"]):
        if df[column].dtype == "object" or pd.api.types.is_string_dtype(df[column]):
            converted_dates = pd.to_datetime(df[column], errors="coerce")

            # Convert only if all non-missing values are valid dates
            original_non_null = df[column].notna().sum()
            if original_non_null > 0 and converted_dates.notna().sum() == original_non_null:
                df[column] = converted_dates

# 4. Convert remaining text columns to categorical data
for column in df.select_dtypes(include=["object", "string"]).columns:
    df[column] = df[column].astype("category")

# 5. Display data types after conversion
print("\nData Types After Conversion:")
print(df.dtypes)

# 6. Display final column types
print("\nFinal Data Type Summary:")
print(df.dtypes.value_counts())

print("\nData types checked and conversion completed.")
```

```python

# Step 9: Rename Columns

# 1. Display original column names
print("Original Column Names:")
print(df.columns.tolist())

# 2. Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# 3. Display renamed column names
print("\nRenamed Column Names:")
print(df.columns.tolist())

print("\nColumn names standardized successfully!")
```

```python

# Step 10: Check Invalid Values (Corrected Code)

import pandas as pd
import numpy as np

# 1. Display numerical summary
print("Numerical Data Summary:")
display(df.describe(include="all"))

# 2. Check negative values in numeric columns
print("\nNegative Values in Numerical Columns:")

for column in df.columns:
    numeric_values = pd.to_numeric(df[column], errors="coerce")
    negative_count = (numeric_values < 0).sum()

    if negative_count > 0:
        print(f"{column}: {negative_count} negative values found")

# 3. Check laptop-specific invalid values
for column in df.columns:
    col = column.lower()

    # Check prices
    if "price" in col:
        numeric_values = pd.to_numeric(df[column], errors="coerce")
        invalid = numeric_values < 0

        print(f"\nInvalid negative prices in {column}: {invalid.sum()}")

        if invalid.any():
            df.loc[invalid, column] = np.nan

    # Check hardware capacity
    elif any(word in col for word in ["ram", "ssd", "hdd", "graphic_card"]):
        numeric_values = pd.to_numeric(df[column], errors="coerce")
        invalid = numeric_values < 0

        print(f"\nInvalid negative values in {column}: {invalid.sum()}")

        if invalid.any():
            df.loc[invalid, column] = np.nan

    # Check weight
    elif "weight" in col:
        numeric_values = pd.to_numeric(df[column], errors="coerce")
        invalid = numeric_values <= 0

        print(f"\nInvalid non-positive weights in {column}: {invalid.sum()}")

        if invalid.any():
            df.loc[invalid, column] = np.nan

# 4. Inspect categorical values
print("\nCategorical Column Values:")

for column in df.select_dtypes(
    include=["object", "category", "string"]
).columns:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False).head(20))

# 5. Check date/time columns
print("\nDate/Time Column Checks:")

date_columns = df.select_dtypes(
    include=["datetime", "datetimetz"]
).columns

if len(date_columns) == 0:
    print("No date/time columns found.")
else:
    for column in date_columns:
        print(f"{column}: {df[column].isna().sum()} missing values")

# 6. Final verification
print("\nTotal Missing Values:", df.isnull().sum().sum())
print("Invalid value checks completed successfully.")
```

```python

# Step 11: Verify the Cleaned Dataset

# 1. Check dataset shape
print("1. DATASET SHAPE")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# 2. Check missing values per column
print("\n2. MISSING VALUES PER COLUMN")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

# 3. Check duplicate rows
print("\n3. DUPLICATE ROWS")
print("Number of duplicate rows:", df.duplicated().sum())

# 4. Check data types
print("\n4. DATA TYPES")
print(df.dtypes)

# 5. Display dataset information
print("\n5. DATASET INFORMATION")
df.info()

# 6. Display descriptive statistics
print("\n6. DESCRIPTIVE STATISTICS")
display(df.describe(include="all"))

# 7. Final verification summary
print("\n7. CLEANING VERIFICATION SUMMARY")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nDataset verification completed!")
```

```python

# Step 13: Mean, Median, Mode, and Standard Deviation

import pandas as pd
import numpy as np

# 1. Select important numerical columns that exist in the dataset
preferred_columns = ["price", "ram_gb", "ssd", "hdd", "graphic_card_gb"]

numeric_columns = df.select_dtypes(include="number").columns.tolist()

selected_columns = [
    col for col in preferred_columns if col in numeric_columns
][:3]

# If preferred columns are unavailable, use the first 3 numeric columns
if len(selected_columns) < 3:
    for col in numeric_columns:
        if col not in selected_columns:
            selected_columns.append(col)
        if len(selected_columns) == 3:
            break

# 2. Calculate statistics for numerical columns
if selected_columns:
    print("Numerical Statistics:")

    for column in selected_columns:
        values = df[column].dropna()

        print(f"\nColumn: {column}")
        if values.empty:
            print("No valid numerical values available.")
            continue

        print("Mean:", values.mean())
        print("Median:", values.median())
        print("Mode:", values.mode().tolist())
        print("Standard Deviation:", values.std())

else:
    print("No numerical columns found.")

# 3. Calculate mode for categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

print("\nMode of Categorical Columns:")

if categorical_columns:
    for column in categorical_columns:
        modes = df[column].mode(dropna=True).tolist()
        print(f"{column}: {modes}")
else:
    print("No categorical columns found.")
```

```python

# Step 14: Correlation Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Select numerical columns
numeric_df = df.select_dtypes(include=["number"])

# 2. Calculate the correlation matrix
correlation_matrix = numeric_df.corr()

print("Correlation Matrix:")
display(correlation_matrix.round(2))

# 3. Plot a correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Matrix of Numerical Columns")
plt.tight_layout()
plt.show()

# 4. Extract unique column pairs (exclude self-correlations)
upper_triangle = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

correlations = upper_triangle.stack().dropna()

# 5. Display strongest positive and negative correlations
if correlations.empty:
    print("Not enough numerical columns with valid data to compare.")
else:
    print("\nStrongest Positive Correlations:")
    positive = correlations[correlations > 0].sort_values(ascending=False)
    if positive.empty:
        print("No positive correlations found.")
    else:
        print(positive.head(5).round(3))

    print("\nStrongest Negative Correlations:")
    negative = correlations[correlations < 0].sort_values()
    if negative.empty:
        print("No negative correlations found.")
    else:
        print(negative.head(5).round(3))
```

```python

# Step 15: Exploratory Data Analysis (EDA)

import pandas as pd
import numpy as np

# 1. Value counts for categorical columns
print("1. VALUE COUNTS FOR CATEGORICAL COLUMNS")

categorical_columns = df.select_dtypes(
    include=["object", "category", "string"]
).columns

for column in categorical_columns:
    print(f"\nValue counts for {column}:")
    display(df[column].value_counts(dropna=False).head(10))

# 2. Highest and lowest values in numerical columns
print("\n2. HIGHEST AND LOWEST VALUES")

numeric_df = df.select_dtypes(include="number")

for column in numeric_df.columns:
    values = numeric_df[column].dropna()

    if not values.empty:
        print(f"\nColumn: {column}")
        print("Minimum value:", values.min())
        print("Maximum value:", values.max())

# 3. Outlier detection using the IQR method
print("\n3. OUTLIER CHECK USING IQR")

for column in numeric_df.columns:
    values = numeric_df[column].dropna()

    if len(values) < 4:
        print(f"{column}: Not enough values for a useful IQR check.")
        continue

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = values[
        (values < lower_limit) | (values > upper_limit)
    ]

    print(f"\nColumn: {column}")
    print("Lower limit:", round(lower_limit, 2))
    print("Upper limit:", round(upper_limit, 2))
    print("Number of outliers:", len(outliers))

# 4. Average numerical values by categorical groups
print("\n4. GROUP AVERAGES")

if len(categorical_columns) > 0 and len(numeric_df.columns) > 0:
    # Use the first categorical column and the first numerical column
    category_column = categorical_columns[0]
    numeric_column = numeric_df.columns[0]

    print(f"\nAverage {numeric_column} by {category_column}:")

    group_averages = df.groupby(
        category_column,
        observed=True
    )[numeric_column].mean().sort_values(ascending=False)

    display(group_averages.head(10).round(2))

else:
    print("Categorical or numerical columns are not available.")

print("\nEDA analysis completed successfully!")
```

```python

# Chart 9: Pair Plot of Laptop Features (Fixed)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Make a copy of the dataset
pair_df = df.copy()

# Convert selected laptop features to numeric values
possible_columns = [
    "price", "ram_gb", "ssd", "hdd", "graphic_card_gb"
]

for col in possible_columns:
    if col in pair_df.columns:
        pair_df[col] = pd.to_numeric(
            pair_df[col].astype(str).str.replace(",", "", regex=False)
            .str.replace("GB", "", regex=False)
            .str.strip(),
            errors="coerce"
        )

# Keep only the selected columns that exist
available_columns = [
    col for col in possible_columns if col in pair_df.columns
]

pair_df = pair_df[available_columns].replace(
    [np.inf, -np.inf], np.nan
).dropna()

# Remove columns with no variation
pair_df = pair_df.loc[:, pair_df.nunique() > 1]

print("Numerical columns available:", pair_df.columns.tolist())
print("Rows available:", len(pair_df))

if len(pair_df.columns) >= 2 and len(pair_df) >= 2:
    pair_sample = pair_df.sample(
        n=min(300, len(pair_df)),
        random_state=42
    )

    g = sns.pairplot(pair_sample, diag_kind="hist", corner=True)
    g.fig.suptitle("Pair Plot of Laptop Features", y=1.02, fontsize=16)
    g.savefig("visualization_09.png", dpi=300, bbox_inches="tight")
    plt.show()

    print("Saved successfully as visualization_09.png")
else:
    print("Still not enough valid numerical columns.")
    print("Your dataset columns are:", df.columns.tolist())
    print("Run print(df.dtypes) and share the output.")
```
