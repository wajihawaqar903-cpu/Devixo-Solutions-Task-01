```python

import pandas as pd

df = pd.read_csv("/content/dataset/laptop_df.csv")

print("Dataset loaded successfully!")
display(df.head())
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Dataset size:", df.size)
print("Column names:", df.columns.tolist())
```

```python

# Check missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())
```

```python

# Check duplicate rows
print("Total duplicate rows:", df.duplicated().sum())
```

```python

df.describe()
```

```python
# Check missing values and percentages
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_summary = missing_summary[missing_summary["Missing Values"] > 0]

display(missing_summary.round(2))
```

```python
print("Specification rating statistics:")
print(df["spec_rating"].describe())

print("\nExisting rating values:")
print(df["spec_rating"].dropna().head(10))

print("\nUnique rating values:")
print(df["spec_rating"].nunique())
```

```python
# Fill missing categorical values with the mode
categorical_columns = ["Ram_type", "ROM_type", "GPU", "OS"]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Categorical missing values filled successfully!")

print(df[categorical_columns].isnull().sum())
```

```python
# Fill missing resolution values with the median
numerical_columns = ["resolution_width", "resolution_height"]

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())

print("Resolution missing values filled successfully!")

print(df[numerical_columns].isnull().sum())
```

```python


# Check ROM and spec_rating before cleaning
print("ROM data type:", df["ROM"].dtype)
print("\nROM sample values:")
print(df["ROM"].value_counts(dropna=False).head(10))

print("\nSpecification rating statistics:")
print(df["spec_rating"].describe())
```

```python
print("ROM values:")
print(df["ROM"].dropna().unique()[:20])

print("\nMissing ROM values:", df["ROM"].isnull().sum())
```

```python
# Fill missing ROM values with the most common value
rom_mode = df["ROM"].mode()[0]

df["ROM"] = df["ROM"].fillna(rom_mode)

print("ROM missing values after cleaning:")
print(df["ROM"].isnull().sum())
```

```python
print("Missing values after current cleaning:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
print("First 10 index values:")
print(df["Unnamed: 0"].head(10).tolist())

print("\nUnique values:", df["Unnamed: 0"].nunique())
print("Minimum value:", df["Unnamed: 0"].min())
print("Maximum value:", df["Unnamed: 0"].max())

print("\nActual DataFrame index:")
print(df.index[:10].tolist())
```

```python
# Remove unnecessary index column
df = df.drop(columns=["Unnamed: 0"])

print("Column removed successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
print("Updated data types:")
print(df.dtypes)
```

```python
print("Dataset shape:", df.shape)

print("\nTotal duplicate rows:", df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDataset information:")
df.info()
```

```python
# Statistical analysis of numerical columns

numerical_columns = [
    "price",
    "spec_rating",
    "display_size",
    "resolution_width",
    "resolution_height",
    "warranty"
]

print("MEAN")
display(df[numerical_columns].mean())

print("MEDIAN")
display(df[numerical_columns].median())

print("MODE")
display(df[numerical_columns].mode().iloc[0])

print("MINIMUM")
display(df[numerical_columns].min())

print("MAXIMUM")
display(df[numerical_columns].max())

print("RANGE")
display(df[numerical_columns].max() - df[numerical_columns].min())

print("STANDARD DEVIATION")
display(df[numerical_columns].std())

print("VARIANCE")
display(df[numerical_columns].var())
```

```python

# Statistical Analysis of Numerical Columns

numerical_columns = [
    "price",
    "spec_rating",
    "display_size",
    "resolution_width",
    "resolution_height",
    "warranty"
]

print("MEAN")
display(df[numerical_columns].mean())

print("MEDIAN")
display(df[numerical_columns].median())

print("MODE")
display(df[numerical_columns].mode().iloc[0])

print("MINIMUM")
display(df[numerical_columns].min())

print("MAXIMUM")
display(df[numerical_columns].max())

print("RANGE")
display(
    df[numerical_columns].max()
    - df[numerical_columns].min()
)

print("STANDARD DEVIATION")
display(df[numerical_columns].std())

print("VARIANCE")
display(df[numerical_columns].var())
```

```python

import matplotlib.pyplot as plt

# Laptop Price Distribution

plt.figure(figsize=(10, 5))

plt.hist(df["price"].dropna(), bins=30, edgecolor="black")

plt.title("Distribution of Laptop Prices")
plt.xlabel("Price")
plt.ylabel("Number of Laptops")

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

# Create graphs for numerical columns

columns = [
    "price",
    "display_size",
    "resolution_width",
    "resolution_height",
    "warranty"
]

for col in columns:
    plt.figure(figsize=(8, 4))
    df[col].dropna().hist(bins=20, edgecolor="black")
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Number of Laptops")
    plt.tight_layout()
    plt.show()
```

```python

# Correlation Analysis

numerical_columns = [
    "price",
    "spec_rating",
    "display_size",
    "resolution_width",
    "resolution_height",
    "warranty"
]

correlation = df[numerical_columns].corr()

print("Correlation Matrix:")
display(correlation)

# Visualize correlation using a heatmap

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Laptop Features")
plt.tight_layout()
plt.show()
```
